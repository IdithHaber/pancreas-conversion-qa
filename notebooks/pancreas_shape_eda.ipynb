## 3. Inspect outputs

- `shape_summary.csv`: volume, bounding dimensions, and PCA spreads per mask.
- `curvature_by_case_and_scale.csv`: per-case surface and principal-curvature summaries in physical units.
- `image_mask_alignment.csv`: geometry checks, populated when CT files are supplied.
- `eda_summary.png`: volume distribution and curvature sensitivity to smoothing scale.
- `median_volume_case_curvature.png`: curvature-colored surface for the case nearest the median mask volume.

The 80 cases are the statistical observations; slices and voxels within a patient are not independent cases. Curvature is sensitive to segmentation boundaries and smoothing scale, so interpret the case distributions alongside the sensitivity plot.
